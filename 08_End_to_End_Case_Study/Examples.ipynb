# End-to-End Analytics Case Study
Retail analytics reporting from raw data to executive summary.
Run the setup, then each demonstration.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

candidates = [Path('data/retail_orders_synthetic.csv'), Path('../data/retail_orders_synthetic.csv'), Path('retail_orders_synthetic.csv')]
source = next((p for p in candidates if p.exists()), None)
if source is None:
    raise FileNotFoundError('Upload retail_orders_synthetic.csv to the Colab session or run from the repository root.')
df = pd.read_csv(source, parse_dates=['order_date'])
df['gross_sales'] = df['units'] * df['unit_price']
df['net_sales'] = df['gross_sales'] * (1 - df['discount_rate'])
df['profit'] = df['net_sales'] - df['cost']
sales = df.loc[df['status'].eq('Completed')].copy()
sales['month'] = sales['order_date'].dt.to_period('M').astype(str)
print('Rows:', len(df), 'Completed:', len(sales))
print(sales[['net_sales','profit']].describe().round(2))
```

## Full analytics report


```python
overview={'completed_orders':len(sales),'revenue':round(sales['net_sales'].sum(),2),'profit':round(sales['profit'].sum(),2),'average_order_value':round(sales['net_sales'].mean(),2)}
print(pd.Series(overview))
print('Top category:',sales.groupby('category')['net_sales'].sum().idxmax())
print('Top region:',sales.groupby('region')['net_sales'].sum().idxmax())
```

## Export a reproducible report


```python
from pathlib import Path
out=Path('outputs');out.mkdir(exist_ok=True)
report=sales.groupby('category',as_index=False).agg(revenue=('net_sales','sum'),profit=('profit','sum'),orders=('order_id','nunique'))
report.to_csv(out/'category_performance.csv',index=False)
print('Saved:',out/'category_performance.csv')
```
