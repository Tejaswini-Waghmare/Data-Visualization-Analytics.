# Business Analytics
Revenue, orders, margins, segments, and monthly trends.
Run the setup, then each demonstration.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

candidates = [Path('data/retail_orders_synthetic.csv'), Path('../data/retail_orders_synthetic.csv'), Path('retail_orders_synthetic.csv')]
source = next((p for p in candidates if p.exists()), None)
if source is None:
    raise FileNotFoundError('Upload retail_orders_synthetic.csv to the Colab session or run from the repository root.')
df = pd.read_csv(source, parse_dates=['order_date'])
df['gross_sales'] = df['units'] * df['unit_price']
df['net_sales'] = df['gross_sales'] * (1 - df['discount_rate'])
df['profit'] = df['net_sales'] - df['cost']
sales = df.loc[df['status'].eq('Completed')].copy()
sales['month'] = sales['order_date'].dt.to_period('M').astype(str)
print('Rows:', len(df), 'Completed:', len(sales))
print(sales[['net_sales','profit']].describe().round(2))
```

## Commercial performance


```python
performance=sales.groupby(['region','channel']).agg(revenue=('net_sales','sum'),profit=('profit','sum'),orders=('order_id','nunique')).reset_index()
performance['margin']=performance['profit']/performance['revenue']
print(performance.sort_values('revenue',ascending=False).round(3))
```

## Pivot heatmap


```python
pivot=sales.pivot_table(index='region',columns='category',values='net_sales',aggfunc='sum',fill_value=0)
fig,ax=plt.subplots(figsize=(8,4));sns.heatmap(pivot,annot=True,fmt='.0f',cmap='Blues',ax=ax)
ax.set(title='Sales by region and category');plt.tight_layout();plt.show()
```
