# Matplotlib
Figures, axes, styling, annotations, and exports.
Run the setup, then each demonstration.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

candidates = [Path('data/retail_orders_synthetic.csv'), Path('../data/retail_orders_synthetic.csv'), Path('retail_orders_synthetic.csv')]
source = next((p for p in candidates if p.exists()), None)
if source is None:
    raise FileNotFoundError('Upload retail_orders_synthetic.csv to the Colab session or run from the repository root.')
df = pd.read_csv(source, parse_dates=['order_date'])
df['gross_sales'] = df['units'] * df['unit_price']
df['net_sales'] = df['gross_sales'] * (1 - df['discount_rate'])
df['profit'] = df['net_sales'] - df['cost']
sales = df.loc[df['status'].eq('Completed')].copy()
sales['month'] = sales['order_date'].dt.to_period('M').astype(str)
print('Rows:', len(df), 'Completed:', len(sales))
print(sales[['net_sales','profit']].describe().round(2))
```

## Subplots


```python
fig,axes=plt.subplots(1,2,figsize=(12,4))
sales.groupby('region')['net_sales'].sum().plot.bar(ax=axes[0],title='Sales by region')
axes[0].set_ylabel('USD')
axes[1].hist(sales['net_sales'],bins=24)
axes[1].set(title='Order value distribution',xlabel='Net sales (USD)')
plt.tight_layout();plt.show()
```

## Annotations


```python
s=sales.groupby('category')['profit'].sum().sort_values()
fig,ax=plt.subplots(figsize=(8,4));s.plot.barh(ax=ax)
ax.set(title='Total profit by category',xlabel='USD');plt.tight_layout();plt.show()
```
