```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

candidates = [Path('data/retail_orders_synthetic.csv'), Path('../data/retail_orders_synthetic.csv'), Path('retail_orders_synthetic.csv')]
source = next((p for p in candidates if p.exists()), None)
if source is None:
    raise FileNotFoundError('Upload retail_orders_synthetic.csv to the Colab session or run from the repository root.')
df = pd.read_csv(source, parse_dates=['order_date'])
df['gross_sales'] = df['units'] * df['unit_price']
df['net_sales'] = df['gross_sales'] * (1 - df['discount_rate'])
df['profit'] = df['net_sales'] - df['cost']
sales = df.loc[df['status'].eq('Completed')].copy()
sales['month'] = sales['order_date'].dt.to_period('M').astype(str)
print('Rows:', len(df), 'Completed:', len(sales))
print(sales[['net_sales','profit']].describe().round(2))
```

## Chart selection and comparison


```python
summary = sales.groupby('category',as_index=False)['net_sales'].sum().sort_values('net_sales',ascending=False)
fig, ax = plt.subplots(figsize=(8,4))
ax.bar(summary['category'],summary['net_sales'])
ax.set(title='Completed sales by category',ylabel='Net sales (USD)',xlabel='Category')
plt.tight_layout(); plt.show()
```

## Time trends


```python
monthly = sales.groupby('month',as_index=False)['net_sales'].sum()
fig,ax=plt.subplots(figsize=(9,4))
ax.plot(monthly['month'],monthly['net_sales'],marker='o')
ax.set(title='Monthly net sales',ylabel='USD',xlabel='Month')
plt.xticks(rotation=45);plt.tight_layout();plt.show()
```
